>**Exception Handling**

```python
a, b = 10, 0
try:
    result = a / b  
    print("Result:", result)
except ZeroDivisionError as e:
    print("Error: Division by zero is not allowed.")
    print("Exception message:", e)
```

```python
li = [1, 2, 3, 4, 5]
try:
    index = int('xyz')
    print(li[index])
except IndexError as e:
    print("Error: Index out of range.")
    print("Exception message:", e)
except ValueError as e:
    print("Error: Value error occurred.")
    print("Exception message:", e)
except Exception as e:
    print("An unexpected error occurred.")
    print("Exception message:", e)
```

```python
li = [1, 2, 3, 4, 5]
try:
    index = int('xyz')
    print(li[index])
except (IndexError, ValueError) as e:
    print("Exception message:", e)
except Exception as e:
    print("An unexpected error occurred.")
    print("Exception message:", e)
```

```python
a = 10
b = 10

try:
    res = a/b
    print(res)
except ZeroDivisionError:
    print('denominator cannot be zero!')
except ValueError:
    print('Value Error Occured!')
else:
    print('There was no error on the code!')
```

```python
def multiply(a, b):
    try:
        return a*b
    except Exception as e:
        raise e
    finally:
        print('End of Program!')

a = 10
b = 10

print(multiply(a, b))

try:
    print(multiply('abc', 'xyz'))
except:
    print('error occured!')
```

`finally` block will be executed for sure!

```python
class NegativeError(Exception):
    def __init__(self):
        self.msg = '-ve dimensions'
    def __str__(self):
        return self.msg

def area(l, w):
    if isinstance(l, (int, float)) and isinstance(w, (int, float)):
        if l < 0 or w < 0:
            raise NegativeError()
        else:
            return l*w
    else:
        raise TypeError('dimensions has to be of numeric types')

print(area(10, -12.3))
```

>**File Handling**

```python
file_name = 'data.txt'
try:
    file = open(file_name)
    content = file.read()
    print(content)
except FileNotFoundError as e:
    print(f"Error: The file '{file_name}' was not found.")
    print("Exception message:", e)
else:
    file.close()
    print("File read successfully.")
```

random access on binary files

```python
# from the start of the file
with open('my.data', 'rb') as file:
    # print(file.read(5).decode())
    file.seek(8, 0)  # Move the cursor to the 6th byte
    print(file.read(10).decode())
    file.seek(0, 0)  # Move the cursor back to the start
    print(file.read(10).decode())
    file.seek(-5, 1)  # Move the cursor back 5 bytes from the current position
    print(file.read(10).decode())

# from the end of the file
with open('my.data', 'rb') as file:
    file.seek(-10, 2)  # Move the cursor to the last 10 bytes
    print(file.read().decode())
```

`seek(offset, from)` here `from=0` means from the beginning of the file, `from=1` means from current position, `from=2` means from the end of the file.
