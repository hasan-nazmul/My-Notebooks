> **Multithreading**

```python
from threading import *

def display():
    for i in range(65, 91):
        print(chr(i))

t = Thread(target=display, name='Alphabets')

t.start()

for i in range(65, 91):
    print(i)

t.join()
```

```python
from threading import *

class Alphabets(Thread):
    def run(self):
        for i in range(65, 91):
            print(chr(i))
            sleep(0.1)

t = Alphabets()

t.start()

for i in range(65, 91):
    print(i)

t.join()
```

Make a class inherited from Thread and override the run method to display alphabets. And also make the task function named `run`.

```python
from threading import *
from time import *

def display(s):
    l.acquire()
    for _ in s:
        print(_)
        sleep(1)
    l.release()

l = Lock()
t1 = Thread(target=display, args=('HELLO WORLD', ))
t2 = Thread(target=display, args=('welcome to programming', ))

t1.start()
t2.start()

t1.join()
t2.join()
```

```python
from threading import *
from time import *

def display(s):
    l.acquire()
    for _ in s:
        print(_)
        sleep(1)
    l.release()

l = Semaphore(2)

t1 = Thread(target=display, args=('HELLO WORLD', ))
t2 = Thread(target=display, args=('welcome to programming', ))
t3 = Thread(target=display, args=('0123456789', ))

t1.start()
t2.start()
t3.start()

t1.join()
t2.join()
t3.join()
```

`semaphore` builds a queue so that one thread after another is allowed to get executed. The amount of threads allowed once at a time can also be mentioned, like `2` was mentioned above!

```python
from threading import *
from time import *

class MyData():
    def __init__(self):
        self.data = 1
        self.flag = False
        self.lock = Lock()

    def puts(self, data):
        while self.flag:
            pass
        self.lock.acquire()
        self.data = data
        self.flag = True
        print('Producer:', data)
        self.lock.release()

    def gets(self):
        while not self.flag:
            pass
        self.lock.acquire()
        x = self.data
        self.flag = False
        self.lock.release()
        return x

def producer(data):
    a = 1
    b = 1
    n = 1
    while True:
        data.puts(n)
        a, b = b, n
        n = a+b
        sleep(1)

def consumer(data):
    while True:
        print('Consumer:', data.gets())
        sleep(1)

md = MyData()

t1 = Thread(target=lambda: producer(md))
t2 = Thread(target=lambda: consumer(md))

t1.start()
t2.start()

t1.join()
t2.join()
```

```python
from threading import *
from time import *

class MyData():
    def __init__(self):
        self.data = 1
        self.cv = Condition()

    def puts(self, data):
        self.cv.acquire()
        self.cv.wait(timeout=0)
        self.data = data
        print('Producer:', data)
        self.cv.notify()
        self.cv.release()

    def gets(self):
        self.cv.acquire()
        self.cv.wait(timeout=0)
        x = self.data
        self.cv.notify()
        self.cv.release()
        return x

def producer(data):
    a = 1
    b = 1
    n = 1
    while True:
        data.puts(n)
        a, b = b, n
        n = a+b
        sleep(1)

def consumer(data):
    while True:
        print('Consumer:', data.gets())
        sleep(1)

md = MyData()

t1 = Thread(target=lambda: producer(md))
t2 = Thread(target=lambda: consumer(md))

t1.start()
t2.start()

t1.join()
t2.join()
```

Both code works same, but here producer writes and notifies to the consumer, waits until cosuming is done. And then, consumer consumes and notifies the producer about ending consumption.
