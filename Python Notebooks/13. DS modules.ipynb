```python
l = [1, 3, 4, 5, 7, 9]
```

```python
import copy as cp
l_copy = cp.copy(l)  # Shallow copy of the list
l_deepcopy = cp.deepcopy(l)  # Deep copy of the list
```

```python
print(id(l), id(l_copy), id(l_deepcopy))  # Check memory addresses
```

```python
print(id(l[0]), id(l_copy[0]), id(l_deepcopy[0]))  # Check memory addresses of elements
print(id(l[1]), id(l_copy[1]), id(l_deepcopy[1]))  # Check memory addresses of elements
```

```python
l = [[1, 2], [3, 4]]
L_copy = cp.copy(l)       # Shallow copy
L_deepcopy = cp.deepcopy(l)  # Deep copy

# Modify the nested list in L_copy
L_copy[0][0] = 99

print(l)          # Output: [[99, 2], [3, 4]] (affected by shallow copy)
print(L_deepcopy) # Output: [[1, 2], [3, 4]]  (unchanged, deep copy is independent)
```

For immutable elements (e.g., `integers`, `strings`), `copy()` and `deepcopy()` behave identically at the element level.

For mutable elements (e.g., nested lists), `deepcopy()` ensures complete independence from the original.

> Inventory

```python
inventory = {
    'apples': 10,
    'bananas': 5,
    'oranges': 8,
    'grapes': 12,
    'pears': 6
}

order = {
    'apples': 3,
    'bananas': 2,
    'pears': 2
}
```

```python
def update_inventory(inventory, order):
    inventory_counter = Counter(inventory)
    order_counter = Counter(order)
    updated_inventory = inventory_counter - order_counter  # Subtract order from inventory
    return dict(updated_inventory)  # Convert back to a regular dictionary
```

```python
update_inventory(inventory, order)  # Call the function to update inventory
```

>Jaccard Similarity

```python
import re

str1 = ('Time is the most valuable thing we have,'
        ' and once lost, it never returns.')

str2 = ("We never get time back once it's "
    "gone—it's truly the most valuable resource.")

words1 = re.findall(r'\w+', str1.lower())

words2 = re.findall(r'\w+', str2.lower())

wset1 = set(words1)
wset2 = set(words2)

common = wset1 & wset2
unique = wset1 | wset2

ratio = len(common) / len(unique)

print(f"Jaccard Similarity:{ratio:.2f}")
```

>Barbarshop

```python
from collections import deque

class Barbarshop:
    def __init__(self):
        self.customers = deque()  # Initialize an empty deque for customers

    def walk_in(self, name):
        self.customers.append(name)

    def serviced(self):
        if self.customers:
            return 'Customer served: ' + self.customers.popleft()  # Serve the first customer in line
        else:
            return None  # No customers to serve
        
    def remaining(self):
        return list(self.customers)
```

```python
Bs = Barbarshop()
Bs.walk_in("Alice")
Bs.walk_in("Bob")
Bs.walk_in("Charlie")
print(Bs.serviced())  # Serve the first customer in line
print(Bs.serviced())  # Serve the next customer in line
print(Bs.remaining())  # Check remaining customers in the queue
```

>Lunch Queue

```python
class LunchQueue:
    def __init__(self, li):
        self.queue = deque(li)  # Initialize an empty deque for the lunch queue

    def join_queue(self, name):
        self.queue.append(name)

    def serve(self, meal):
        service_line = f'{meal} Served: '
        for _ in self.queue:
            service_line += _ + ' '
        self.queue.rotate(-1)  # Rotate the queue to the left
        return service_line
```

```python
Lq = LunchQueue(['Alice', 'Bob', 'Charlie'])
Lq.join_queue('David')
print(Lq.serve('Breakfast'))  # Serve the first customer in line
print(Lq.serve('Lunch'))  # Serve the next customer in line
print(Lq.serve('Dinner'))  # Serve the next customer in line
```

**statistics module**

```python
data = [1, 2, 3, 1, 2, 2, 1, 4, 5, 6, 7, 8, 9, 10, 1, 3, 2, 1, 4, 5, 6, 7, 8, 9, 10]
```

```python
import statistics as stat
mean = stat.mean(data)  # Calculate the mean
median = stat.median(data)  # Calculate the median
mode = stat.mode(data)  # Calculate the mode
media_high = stat.median_high(data)  # Calculate the median high
media_low = stat.median_low(data)  # Calculate the median low
print(f"Mean: {mean}, Median: {median}, Mode: {mode}, Median High: {media_high}, Median Low: {media_low}")
```

```python
# For entire list's standard deviation and variance
std_dev = stat.pstdev(data)  # Calculate the standard deviation
variance = stat.pvariance(data)  # Calculate the variance
print(f"Standard Deviation: {std_dev}, Variance: {variance}")  # Print the standard deviation and variance
```
