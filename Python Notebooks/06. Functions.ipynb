```python
def PascalsTriangle(i, n, li=None):
    if i > n:
        return
    print(' '*(n-i), end='')
    if li is None:
        li = [1]
    else:
        if len(li) == 1:
            li = [1, 1]
        else:
            new_li = [1]
            for j in range(1, len(li)):
                new_li.append(li[j]+li[j-1])
            new_li.append(1)
            li = new_li
    for j in li:
        print(j, end=' ')
    print()
    PascalsTriangle(i+1, n, li)

PascalsTriangle(1, 5)
```

```python
def PascalsTriangle(n):
    r = [1]
    for i in range(n):
        print((n-i-1)*' ', end='')
        for i in r:
            print(i, end=' ')
        print()
        tr = [0] + r
        r = r + [0]
        r = [x+y for x,y in zip(tr, r)]

PascalsTriangle(5)
```

>**Decorator**

```python
def decorator(f):
    def Inner():
        print('+' * 15)
        f()
        print('+' * 15)
    return Inner

def display():
    msg = 'WELCOME'
    print(f'{msg:-^15}')

display = decorator(display)

display()
```

The definition of display has been changed.

```python
def decorator(f):
    def Inner():
        print('+' * 15)
        f()
        print('+' * 15)
    return Inner

@decorator
def display():
    msg = 'WELCOME'
    print(f'{msg:-^15}')

display()
```

```python
def requires_login(msg):
    def decorator(func):
        def wrapper(user, *args, **kwargs):
            if not user.get('logged_in', False):
                return msg
            return func(user, *args, **kwargs)
        return wrapper
    return decorator

@requires_login(msg="You must be logged in to view your profile.")
def view_profile(user):
    return f"Profile: {user['name']}"

@requires_login(msg="You must be logged in to delete your account.")
def delete_account(user):
    return f"Account {user['name']} deleted"

# Test with logged out user
guest = {'name': 'Guest', 'logged_in': False}
print(view_profile(guest))  # Access denied. Please log in first.

# Test with logged in user
alice = {'name': 'Alice', 'logged_in': True}
print(view_profile(alice))  # Profile: Alice
```

Layers when using @requires_login(msg=...):
1) requires_login(msg): outer factory — captures the message and returns decorator(func).
2) decorator(func): decoration-time wrapper — takes the original function and returns wrapper.
3) wrapper(user, *args, **kwargs): call-time wrapper — checks user['logged_in'];
   returns the msg if not logged in, otherwise calls the original function.

Decoration flow: requires_login(msg) -> decorator -> wrapper replaces the original function.

Call with guest -> You must be logged in to view your profile.
Call with alice -> Profile: Alice

```python
print("\n=== Common Function Pitfalls ===\n")

# ❌ PITFALL 1: Mutable default arguments
print("❌ Pitfall 1: Mutable Default Arguments")

def bad_append(item, items=[]):
    items.append(item)
    return items

list1 = bad_append(1)
list2 = bad_append(2)
print(f"  list1 = {list1}  (Expected: [1])")
print(f"  list2 = {list2}  (Expected: [2])")
print("  ✗ Both share the same list!")

# ✓ SOLUTION
def good_append(item, items=None):
    if items is None:
        items = []
    items.append(item)
    return items

list1 = good_append(1)
list2 = good_append(2)
print(f"\n✓ Solution:")
print(f"  list1 = {list1}")
print(f"  list2 = {list2}")

# ❌ PITFALL 2: Closure variable binding
print("\n\n❌ Pitfall 2: Late Binding in Closures")

functions = []
for i in range(3):
    functions.append(lambda x: x + i)

print(f"  Function results: {[f(10) for f in functions]}")
print("  ✗ All use i=2 (late binding)")

# ✓ SOLUTION
functions = []
for i in range(3):
    functions.append(lambda x, i=i: x + i)  # Default captures current i

print(f"\n✓ Solution: {[f(10) for f in functions]}")

# ❌ PITFALL 3: Global variable modification
print("\n\n❌ Pitfall 3: Modifying Global State")

counter = 0

def bad_increment():
    global counter
    counter += 1  # Modifies global state
    return counter

print(f"  counter = {bad_increment()}")
print(f"  counter = {bad_increment()}")
print("  ✗ Hard to test, side effects")

# ✓ SOLUTION
def good_increment(count):
    return count + 1

state = 0
state = good_increment(state)
state = good_increment(state)
print(f"\n✓ Pure function: No side effects")
print(f"  state = {state}")
```

```python
from functools import partial

print("=== Partial Functions & Currying ===\n")

# Original function
def multiply(x, y, z):
    return x * y * z

# Using partial to fix parameters
multiply_by_2 = partial(multiply, 2)
multiply_by_2_and_3 = partial(multiply, 2, 3)

print("Original: multiply(x, y, z)")
print(f"  multiply(2, 3, 4) = {multiply(2, 3, 4)}")

print("\nPartial application:")
print(f"  multiply_by_2 = partial(multiply, 2)")
print(f"  multiply_by_2(3, 4) = {multiply_by_2(3, 4)}")

print(f"\n  multiply_by_2_and_3 = partial(multiply, 2, 3)")
print(f"  multiply_by_2_and_3(4) = {multiply_by_2_and_3(4)}")

# Real-world example: API client
def make_request(base_url, endpoint, method='GET', timeout=10):
    return f"{method} {base_url}{endpoint} (timeout: {timeout}s)"

# Create specialized clients
github_api = partial(make_request, 'https://api.github.com')
github_users = partial(github_api, '/users')

print("\n=== Real-world API example ===")
print(f"Request: {github_users('/octocat')}")

# Currying for data transformation
def curry(f):
    """Convert function to curried version"""
    from functools import wraps
    
    @wraps(f)
    def curried(*args, **kwargs):
        if len(args) + len(kwargs) >= f.__code__.co_argcount:
            return f(*args, **kwargs)
        return partial(curried, *args, **kwargs)
    
    return curried

@curry
def add(a, b, c):
    return a + b + c

print("\n=== Currying example ===")
print(f"add(1)(2)(3) = {add(1)(2)(3)}")
print(f"add(1, 2)(3) = {add(1, 2)(3)}")
print(f"add_1 = add(1)")
print(f"add_1_2 = add_1(2)")
print(f"add_1_2(3) = {add_1(2)(3)}")
```

```python
print("\n=== Function Composition & Pipelines ===\n")

# Basic composition
def compose(*functions):
    """Compose functions right-to-left"""
    def composed(x):
        for func in reversed(functions):
            x = func(x)
        return x
    return composed

def compose_left(*functions):
    """Compose functions left-to-right (pipeline)"""
    def composed(x):
        for func in functions:
            x = func(x)
        return x
    return composed

# Example functions
def double(x):
    return x * 2

def add_ten(x):
    return x + 10

def square(x):
    return x ** 2

# Right-to-left composition: f(g(h(x)))
composed_right = compose(square, add_ten, double)
print(f"Right composition: double → add_ten → square")
print(f"  Input: 5")
print(f"  Step 1 (double): 10")
print(f"  Step 2 (add_ten): 20")
print(f"  Step 3 (square): 400")
print(f"  Result: {composed_right(5)}")

# Left-to-right pipeline
pipeline = compose_left(double, add_ten, square)
print(f"\nLeft pipeline: double → add_ten → square")
print(f"  Input: 5")
print(f"  Result: {pipeline(5)}")

# Data processing pipeline
def data_pipeline(data):
    """Process data through multiple stages"""
    steps = [
        ('Parse', lambda x: x.split(',')),
        ('Strip', lambda x: [item.strip() for item in x]),
        ('Filter', lambda x: [item for item in x if item]),
        ('Convert', lambda x: [int(item) if item.isdigit() else item for item in x]),
    ]
    
    print("Data Pipeline Processing:")
    print(f"  Input: '{data}'")
    
    result = data
    for name, func in steps:
        result = func(result)
        print(f"  → {name}: {result}")
    
    return result

final = data_pipeline("1, 2, , 3, 4")
print(f"  Final: {final}\n")
```

```python
from functools import wraps
import time
import random

print("\n=== Error Handling & Retry Patterns ===\n")

def retry(max_attempts=3, delay=1, backoff=2):
    """Decorator for retry with exponential backoff"""
    def decorator(func):
        @wraps(func)
        def wrapper(*args, **kwargs):
            attempt = 1
            while attempt <= max_attempts:
                try:
                    return func(*args, **kwargs)
                except Exception as e:
                    if attempt == max_attempts:
                        raise
                    wait_time = delay * (backoff ** (attempt - 1))
                    print(f"  Attempt {attempt} failed: {e}. Retrying in {wait_time}s...")
                    time.sleep(wait_time)
                    attempt += 1
        return wrapper
    return decorator

@retry(max_attempts=3, delay=0.5, backoff=2)
def unstable_api_call():
    """Simulates an unstable API that fails sometimes"""
    if random.random() < 0.7:  # 70% failure rate
        raise ConnectionError("API temporarily unavailable")
    return "Success!"

print("Calling unstable API with retry mechanism:")
try:
    result = unstable_api_call()
    print(f"✓ Result: {result}")
except Exception as e:
    print(f"✗ Final error: {e}")

# Catch specific exceptions
def handle_multiple_errors(func):
    @wraps(func)
    def wrapper(*args, **kwargs):
        try:
            return func(*args, **kwargs)
        except ValueError as ve:
            print(f"✗ Value error: {ve}")
            return None
        except TypeError as te:
            print(f"✗ Type error: {te}")
            return None
        except Exception as e:
            print(f"✗ Unexpected error: {e}")
            return None
    return wrapper

@handle_multiple_errors
def process_data(data):
    if not isinstance(data, dict):
        raise TypeError("Data must be a dictionary")
    if 'value' not in data:
        raise ValueError("Missing 'value' key")
    return data['value'] * 2

print("\n✓ Testing error handling:")
print(f"Result: {process_data({'value': 5})}")  # Success
print(f"Result: {process_data(None)}")  # TypeError
print(f"Result: {process_data({})}")  # ValueError
```

```python
from functools import lru_cache, wraps
import time

print("=== Function Caching for Performance ===\n")

# Without caching - slow
def fibonacci_slow(n):
    if n < 2:
        return n
    return fibonacci_slow(n-1) + fibonacci_slow(n-2)

# With LRU Cache - fast
@lru_cache(maxsize=128)
def fibonacci_fast(n):
    if n < 2:
        return n
    return fibonacci_fast(n-1) + fibonacci_fast(n-2)

# Benchmark
start = time.time()
result_slow = fibonacci_slow(30)
slow_time = time.time() - start

start = time.time()
result_fast = fibonacci_fast(30)
fast_time = time.time() - start

print(f"Without cache (n=30): {slow_time:.4f}s")
print(f"With cache (n=30): {fast_time:.6f}s")
print(f"Speedup: {slow_time/fast_time:.0f}x faster ⚡")

# Check cache info
print(f"\nCache info: {fibonacci_fast.cache_info()}")

# Custom caching decorator
def cache_with_ttl(ttl_seconds=60):
    def decorator(func):
        cache = {}
        cache_time = {}
        
        @wraps(func)
        def wrapper(*args, **kwargs):
            key = (args, tuple(sorted(kwargs.items())))
            current_time = time.time()
            
            if key in cache and (current_time - cache_time[key]) < ttl_seconds:
                return cache[key]
            
            result = func(*args, **kwargs)
            cache[key] = result
            cache_time[key] = current_time
            return result
        
        return wrapper
    return decorator

@cache_with_ttl(ttl_seconds=60)
def expensive_computation(x):
    time.sleep(0.1)  # Simulate expensive operation
    return x ** 2

print("\n✓ Custom TTL cache: calls within 60s use cached result")
```
