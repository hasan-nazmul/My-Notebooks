```python
from tkinter import *

win = Tk()

win.title('My first APK')

win.geometry('600x400+500+200')
'''
window size is 600x400 
and it is away 500px in X-axis from origin 
and 200px in Y-axis
+/- can be used to place the window
from the top or the bottom
'''

win.resizable(False, True)
# It's possible to resize the window in X axis, not in Y axis

l = Label(win, text='Hello World')
l.pack()
'''
Label has been added, parent window is `win`
l.pack() adds that label to the window
'''

b = Button(win, text='Click Me')
b.pack()
'''
Button adding
'''

e = Entry()
e.pack()
'''
Entry
'''

t = Text(win, width=30, height=10)
t.pack()
'''
Text
'''

c = Checkbutton(win, text='Yes')
c.pack()

y = Radiobutton(win, text='Yes', variable='v1', value=1)
n = Radiobutton(win, text='No', variable='v1', value=0)
y.pack()
n.pack()
'''
variables are same for y and n that's why
only one option can be selected
'''

v = StringVar()
o = OptionMenu(win, v, 'Python', *('Java', 'JS', 'C', 'Cpp'))
o.pack()

s = Scale(win, from_=0, to=100)
s.pack()

win.attributes('-alpha', 0.90)
# -alpha is the opacity of the window

win.mainloop()
```

```python
from tkinter import *
from tkinter.messagebox import *

s = str()

def fun(e):
    global s
    print(s)
    s = ''

def fun1(e):
    global s
    s += e.char

win = Tk()
win.title('My first APK')
win.geometry('600x400')

e = Entry(win, bg='red', fg='yellow')
e.place(x=100, y=100, width=200, height=50)

e.bind('<Button-1>', fun)
e.bind('<KeyPress>', fun1)

win.mainloop()
```

```python
from tkinter import *
from tkinter.messagebox import *

def myhandler():
    lbl['text'] = cb1['text'] if var1.get() else ''
    lbl['text'] += cb2['text'] if var2.get() else ''
    lbl['text'] += cb3['text'] if var3.get() else ''

    if int(var1.get())+int(var2.get())+int(var3.get())==3:
        btn['text'] = 'deselect all'

def newHandler():
    if btn['text'] == 'select all':
        if not var1.get():
            cb1.invoke()
        if not var2.get():
            cb2.invoke()
        if not var3.get():
            cb3.invoke()
        btn['text'] = 'deselect all'
    else:
        if var1.get():
            cb1.invoke()
        if var2.get():
            cb2.invoke()
        if var3.get():
            cb3.invoke()
        btn['text'] = 'select all'

win = Tk()
win.title('Awesome APK')
win.geometry('600x400')

lbl = Label(win, text='Label')
lbl.pack()

var1 = IntVar()
cb1 = Checkbutton(win, variable=var1, text='Java', command=myhandler)
cb1.pack()

var2 = IntVar()
cb2 = Checkbutton(win, variable=var2, text='C++', command=myhandler)
cb2.pack()

var3 = IntVar()
cb3 = Checkbutton(win, variable=var3, text='Python', command=myhandler)
cb3.pack()

btn = Button(win, text='select all', command=newHandler)
btn.pack()

win.mainloop()
```

```python
from tkinter import *
from tkinter.messagebox import *

def up():
    var.set(var.get()+1)
    if int(var.get()) > 0:
        lbl['fg'] = 'green'
    elif not int(var.get()):
        lbl['fg'] = 'black'

def down():
    var.set(var.get()-1)
    if int(var.get()) < 0:
        lbl['fg'] = 'red'
    elif not int(var.get()):
        lbl['fg'] = 'black'

win = Tk()
win.title('Awesome APK')
win.geometry('600x400')

var = IntVar(value=0)
lbl = Label(win, textvariable=var, font=('Arial', 50), width=5)
lbl.grid(row=1, column=1, pady=10, rowspan=2)

plus = Button(win, text='+', font=('Arial', 15), width=3, bg='green', fg='white', command=up)
plus.grid(row=1, column=0, padx=5, rowspan=2)

minus = Button(win, text='-', font=('Arial', 15), width=3, bg='red', fg='white', command=down)
minus.grid(row=1, column=2, padx=5, rowspan=2)

win.mainloop()
```

```python
from tkinter import *
from tkinter.messagebox import *

win = Tk()
win.title('Notepad')
win.geometry('600x400')

txt = Text(win, undo=True, height=10)
txt.grid(row=0, column=1, columnspan=5)

undo = Button(win, text='undo', width=10, pady=10, bg='blue', fg='white', command=lambda : txt.edit_undo())
redo = Button(win, text='redo', width=10, pady=10, bg='yellow', fg='black', command=lambda : txt.edit_redo())
clear_all = Button(win, text='clear all', width=10, pady=10, command=lambda : txt.delete('1.0', END))

undo.grid(row=1, column=2, padx=5, pady=5)
redo.grid(row=1, column=3, padx=5, pady=5)
clear_all.grid(row=1, column=4, padx=5, pady=5)

win.mainloop()
```

```python
from tkinter import *

def myhandler():
    txt1.insert(1.0, 'Hello World')

def pophandler(e):
    file.tk_popup(e.x_root, e.y_root)

win = Tk()
win.geometry('800x600')

txt1 = Text(win)
txt1.pack(fill=BOTH)

menubar = Menu()
win['menu'] = menubar

file = Menu(menubar)
menubar.add_cascade(label='File', menu=file)

file.add_command(label='New', command=myhandler)
file.add_command(label='Open')
file.add(itemType='command', label='Save')
file.add_separator()
file.add_checkbutton(label='Save as', onvalue=1, offvalue=0)

rad1 = Menu(file)
rad1.add_radiobutton(label='option 1')
rad1.add_radiobutton(label='option 2')
rad1.add_radiobutton(label='option 3')
file.add_cascade(label='Options', menu=rad1)

txt1.bind('<Button-3>', pophandler)

win.mainloop()
```

```python
from tkinter import *
from datetime import *

win = Tk()
win.title('Digital Clock')
win.geometry('800x600')

class DigitalClock:
    def __init__(self, win, x, y):
        self.label = Label(win, font=('Times new roman', 20), bg='green', fg='purple')
        self.label.place(x=x, y=y)

    def ticking(self):
        self.label['text'] = datetime.now().strftime('%I:%M:%S')
        self.label.after(100, self.ticking)

dc1 = DigitalClock(win=win, x=10, y=10)
dc2 = DigitalClock(win=win, x=200, y=200)

dc1.ticking()
dc2.ticking()

win.mainloop()
```

```python
from tkinter import *
import time

class Stopwatch:
    def __init__(self, win):
        self.timer = Label(win, text='00:00:000', font=('times new roman', 45), bg='black', fg='white')
        self.start = Button(win, text='start', bg='green', fg='white', command=self.start_handler)
        self.stop = Button(win, text='stop', bg='red', fg='white', command=self.stop_handler)
        self.reset = Button(win, text='reset', bg='blue', fg='white', command=self.reset_handler)
        self.flag = True
        self.elapsed_time = 0
        self.starting_time = 0
        self.timer.grid(row=1, column=1, columnspan=3, padx=5, pady=5)
        self.start.grid(row=2, column=1, padx=5, pady=5)
        self.stop.grid(row=2, column=2, padx=5, pady=5)
        self.reset.grid(row=2, column=3, padx=5, pady=5)

    def update_watch(self):
        inc = int(self.flag) * int(round((time.time() - self.starting_time)*1000))
        self.starting_time = time.time()
        tm = self.elapsed_time
        self.elapsed_time += inc
        mn, sc, ms = (tm // 60000), (tm % 60000) // 1000, (tm % 1000)
        self.timer['text'] = f'{mn:0>2}:{sc:0>2}:{ms:0>3}'
        self.timer.after(10, self.update_watch)

    def start_handler(self):
        self.starting_time = time.time()
        self.flag = True
        self.update_watch()

    def stop_handler(self):
        self.flag = False

    def reset_handler(self):
        self.elapsed_time = 0
        self.flag = False

win = Tk()
win.title('Stopwatch')
win.geometry('800x600')

sw = Stopwatch(win)

win.mainloop()
```
