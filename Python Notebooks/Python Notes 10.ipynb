```python
class Rectangle:
    count = 0
    def __init__(self, length=1, breadth=1):
        if not (isinstance(length, (int, float)) and isinstance(breadth, (int, float))):
            raise TypeError('Dimensions must be of numeric types.')
        if length < 0 or breadth < 0:
            raise ValueError('Dimensions has to be non-negative values!')
        self.length = length
        self.breadth = breadth
        Rectangle.count += 1

    def __str__(self):
        return f'Length: {self.length}\nBreadth: {self.breadth}'

    def area(self):
        return self.length * self.breadth

    def perimeter(self):
        return 2 * (self.length+self.breadth)

    def get_length_n_breadth(self):
        return (id(self.length), id(self.breadth))

    @classmethod
    def get_count(cls):
        return id(cls.count)

r = Rectangle(10, 25)
r1 = Rectangle(35, 40)
r2 = Rectangle(50, 10)
print(Rectangle.get_count(), r.get_count(), r1.get_count(), r2.get_count())
print(r.get_length_n_breadth())
print(r1.get_length_n_breadth())
print(r2.get_length_n_breadth())
```

`count` is not an instance, rather it is a class object. That means all the objects will have the same count variable. But `length`, `breadth` or the methods: `area(self)`, `perimeter(self)` are instance variables and instance methods. Means each object will have its own `length`, `breadth` and `area(self)`, `perimeter(self)`. Thatswhy its possible to write `Rectangle.get_count()` not `Rectangle.area()` or `Rectangle.perimeter()`. `@classmethod` decorator needs to be mentioned before every classmethods each. In `get_count(cls)`, `cls` can have a class as a parameter.

```python
class Rectangle:
    def __init__(self, length=1, breadth=1):
        self.length = length
        self.breadth = breadth

    def __str__(self):
        return f'Length: {self.length}\nBreadth: {self.breadth}'

    def area(self):
        return self.length * self.breadth

    def perimeter(self):
        return 2 * (self.length+self.breadth)

    @staticmethod
    def calc_area(length, breadth):
        return length * breadth

r = Rectangle.calc_area(10, 15)
print(r)
```

The `calc_area(length, breadth)` are like utility methods/static methods. We don't want to create an object, just want to get a service. `@staticmethod` decorator needs to be mentioned before static method. 

By default any method treats the first argument as the instance of the object.

```python
class Rectangle:
    def __init__(self, length=1, breadth=1):
        self.length = length
        self.breadth = breadth

    def __str__(self):
        return f'Length: {self.length}\nBreadth: {self.breadth}'

    @property
    def length(self):
        return self._length

    @length.setter
    def length(self, length):
        self._length = max(length, 1)

r = Rectangle(-10, 15)
r.length = -5
print(r.length)
```

`_length` means protected. `@property` decorator is for getter and `@length.setter` is for setter. Now for direct accessing these getter and setter methods are called, no need for methods for this.

> **Inheritance**

```python
class Rectangle:
    def __init__(self, length=1, breadth=1):
        self.length = length
        self.breadth = breadth

    def area(self):
        return self.length * self.breadth

    def perimeter(self):
        return 2 * (self.length+self.breadth)

class Cuboid(Rectangle):
    def __init__(self, length=1, breadth=1, height=1):
        super().__init__(length, breadth)
        self.height = height

    def volume(self):
        return self.length * self.breadth * self.height

c = Cuboid(10, 10, 10)

print(c.area(), c.perimeter(), c.volume())
```

`Cuboid` inherits from `Rectangle` and the constructor of `Cuboid` needs to call `super().__init__()` for the constructor of its base class.

```python
class Parent:
    def __init__(self, name, age, dob):
        self.name = name
        self.__age = age
        self._dob = dob

    def greet(self):
        print(f'Parent Class: Hello, {self.name}! Your age is {self.__age} and your date of birth is {self._dob}.')

class Child(Parent):
    def __init__(self, name, age, dob, school):
        super().__init__(name, age, dob)
        self.school = school

    def greet(self):
        super().greet()
        print(f'Child Class: Hello, {self.name}! Your date of birth is {self._dob} and you go to {self.school}.')

c = Child('Alice', 10, '2013-05-15', 'Greenwood High')
c.greet()
```

```python
# name mangling
class Parent:
    def __init__(self, name):
        self.name = name

    def __private_method(self):
        return f'Private method in Parent: {self.name}'

    def public_method(self):
        return self.__private_method()

class Child(Parent):
    def __private_method(self):
        return f'Private method in Child: {self.name}'

    def public_method(self):
        return self.__private_method()
    
c = Child('Bob')
print(c.public_method())  # This will call the Child's private method due to name mangling
print(c._Parent__private_method())  # Accessing Parent's private method using name mangling
```

> **Operator Overloading**

```python
class Vector:
    def __init__(self, x, y):
        self.x = x
        self.y = y

    def __add__(self, other):
        x = self.x + other.x
        y = self.y + other.y
        return Vector(x, y)

    def __str__(self):
        return f'x: {self.x}, y: {self.y}'

a = Vector(10, 20)
b = Vector(12, 45)
print(a+b)
```

> **Abstract Method and Abstract Class**

```python
from abc import ABC, abstractmethod

class Parent(ABC):
    def method1(self):
        print('Parent method -- 1')

    @abstractmethod
    def method2(self):
        pass

class Child(Parent):
    def method2(self):
        print('Child (abstract) method -- 2')
    def method3(self):
        print('Child method -- 3')

c = Child()
c.method1()
c.method2()
c.method3()
```

Abstract classes inherits from `ABC` class of `abc` module. And decorator `@abstractmethod` defines an abstract method. Use of `@abstractmethod` forces child class to override the function.

```python
class MinimumBalanceError(Exception):
    pass

class Account:
    acc_number = 1001

    def __init__(self, name='John Doe', balance=5000):
        self.name = name
        self.acc = Account.acc_number

        try:
            self.balance = balance
        except Exception as e:
            print(e)
        else:
            print('Account has been created successfully!')

        Account.acc_number += 1

    @property
    def balance(self):
        return self._balance

    @balance.setter
    def balance(self, balance):
        if not isinstance(balance, (int, float)):
            raise ValueError('Balance has to be of numeric value!')
        if balance < 5000:
            raise MinimumBalanceError('Balance cannot be lesser than 5000!')
        self._balance = balance

    def deposit(self, amount):
        try:
            self.balance += amount
        except Exception as e:
            print(e)
            print(f'Deposit was unsuccessful!')
        else:
            print(f'Deposit of ${amount} to {self.name} was successful!')

    def withdraw(self, amount):
        try:
            self.balance -= amount
        except Exception as e:
            print(e)
            print(f'Withdrawal of {amount} was unsuccessful!')
        else:
            print(f'Withdrawal of ${amount} from {self.name} was successful!')

    def __str__(self):
        return f'Account No: {self.acc}, Name: {self.name}, Balance: ${self.balance}'

a = Account('Nazmul', 12000)
a.deposit(7000)
print(a)
```

```python
import math
from abc import ABC, abstractmethod

class TriangleFormationError(Exception):
    pass

class Polygon(ABC):
    def __init__(self, no_of_sides, *sides):
        self.no_of_sides = no_of_sides
        self.sides = sides
    @abstractmethod
    def Area(self):
        pass

class Triangle(Polygon):
    def __init__(self, *sides):
        super().__init__(3, *sides)
    def Area(self):
        s = sum(self.sides) / 2
        area = math.sqrt(s * math.prod([(s-x) for x in self.sides]))
        return area
    @property
    def sides(self):
        return self._sides
    @sides.setter
    def sides(self, sides):
        if len(sides) != 3:
            raise TriangleFormationError('Triangle must have exactly 3 sides!')
        s = sum(sides)
        for x in sides:
            if not isinstance(x, (int, float)) or x <= 0 or s-2*x < 0:
                raise TriangleFormationError('Triangle Forming is not possible with these sides!')
        self._sides = sides

try:
    t = Triangle(5, 4, 10)
    print(t.sides)
except Exception as e:
    print(e)

try:
    t = Triangle(3, 5, 4)
    print(t.sides)
    print(t.Area())
except Exception as e:
    print(e)
```

```python
class Course:
    class Book:
        def __init__(self, title):
            self.title = title

        def __str__(self):
            return self.title

    def __init__(self, course_title, course_duration, *books):
        self.course_title = course_title
        self.course_duration = course_duration
        self.books = [Course.Book(book) for book in books if isinstance(book, str)]

    def show_details(self):
        print(f'Course Title: {self.course_title}')
        print(f'Course Duration: {self.course_duration}')
        print('Books:')
        sl = 1
        for book in self.books:
            print(f'{sl}. {book}')
            sl += 1

book_list = [
                'The Lord of the Rings', 'Harry Potter', 'The Shining',
                '2001: A Space Odyssey', 'The Godfather',
            ]

c = Course('English Literature', '6 months', *book_list)

c.show_details()
```
