```python
class Rectangle:
    count = 0
    def __init__(self, length=1, breadth=1):
        if not (isinstance(length, (int, float)) and isinstance(breadth, (int, float))):
            raise TypeError('Dimensions must be of numeric types.')
        if length < 0 or breadth < 0:
            raise ValueError('Dimensions has to be non-negative values!')
        self.length = length
        self.breadth = breadth
        Rectangle.count += 1

    def __str__(self):
        return f'Length: {self.length}\nBreadth: {self.breadth}'

    def area(self):
        return self.length * self.breadth

    def perimeter(self):
        return 2 * (self.length+self.breadth)

    def get_length_n_breadth(self):
        return (id(self.length), id(self.breadth))

    @classmethod
    def get_count(cls):
        return id(cls.count)

r = Rectangle(10, 25)
r1 = Rectangle(35, 40)
r2 = Rectangle(50, 10)
print(Rectangle.get_count(), r.get_count(), r1.get_count(), r2.get_count())
print(r.get_length_n_breadth())
print(r1.get_length_n_breadth())
print(r2.get_length_n_breadth())
```

`count` is not an instance, rather it is a class object. That means all the objects will have the same count variable. But `length`, `breadth` or the methods: `area(self)`, `perimeter(self)` are instance variables and instance methods. Means each object will have its own `length`, `breadth` and `area(self)`, `perimeter(self)`. Thatswhy its possible to write `Rectangle.get_count()` not `Rectangle.area()` or `Rectangle.perimeter()`. `@classmethod` decorator needs to be mentioned before every classmethods each. In `get_count(cls)`, `cls` can have a class as a parameter.

```python
class Rectangle:
    def __init__(self, length=1, breadth=1):
        self.length = length
        self.breadth = breadth

    def __str__(self):
        return f'Length: {self.length}\nBreadth: {self.breadth}'

    def area(self):
        return self.length * self.breadth

    def perimeter(self):
        return 2 * (self.length+self.breadth)

    @staticmethod
    def calc_area(length, breadth):
        return length * breadth

r = Rectangle.calc_area(10, 15)
print(r)
```

The `calc_area(length, breadth)` are like utility methods/static methods. We don't want to create an object, just want to get a service. `@staticmethod` decorator needs to be mentioned before static method. 

By default any method treats the first argument as the instance of the object.

```python
class Rectangle:
    def __init__(self, length=1, breadth=1):
        self.length = length
        self.breadth = breadth

    def __str__(self):
        return f'Length: {self.length}\nBreadth: {self.breadth}'

    @property
    def length(self):
        return self._length

    @length.setter
    def length(self, length):
        self._length = max(length, 1)

r = Rectangle(-10, 15)
r.length = -5
print(r.length)
```

`_length` means protected. `@property` decorator is for getter and `@length.setter` is for setter. Now for direct accessing these getter and setter methods are called, no need for methods for this.
