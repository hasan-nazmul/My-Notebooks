```python
from bs4 import BeautifulSoup
import requests

with open('index.html') as html_file:
    soup = BeautifulSoup(html_file, 'lxml')

for article in soup.find_all('article', class_='book'):
    print(article.h3.text)
```

```python
from bs4 import BeautifulSoup
import requests

url = 'https://www.rokomari.com/search?term=%E0%A6%B8%E0%A6%BE%E0%A6%87%E0%A6%AE%E0%A7%81%E0%A6%AE&search_type=ALL&page=1'

source = requests.get(url).text

soup = BeautifulSoup(source, 'lxml')

for item in soup.find_all('h4', class_='book-title'):
    print(item.text)
```

```python
from bs4 import BeautifulSoup
import requests

url = 'https://www.rokomari.com/search?term=saimum+series&search_type=ALL&page=1'

source = requests.get(url).text

soup = BeautifulSoup(source, 'lxml')

for item in soup.find_all('div', class_='book-img'):
    print(item.img['alt'])
```
