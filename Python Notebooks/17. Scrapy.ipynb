```python
import scrapy


class BookspiderSpider(scrapy.Spider):
    name = "bookspider"
    allowed_domains = ["books.toscrape.com"]
    start_urls = ["https://books.toscrape.com"]

    def parse(self, response):
        books = response.css("article.product_pod")

        for book in books:
            yield {
                "title": book.css("h3 a::text").get(),
                "price": book.css(".product_price .price_color::text").get(),
                "url": book.css("h3 a::attr(href)").get(),
            }

        next_page = response.css("li.next a::attr(href)").get()

        if next_page:
            yield response.follow(next_page, self.parse)
```

```python
category = response.xpath('//ul[@class="breadcrumb"]/li[@class="active"]/preceding-sibling::li[1]/a/text()').get()
```

XPath Breakdown:
1. `//ul[@class="breadcrumb"]`
`//` - Search anywhere in the document
`ul[@class="breadcrumb"]` - Find a <ul> element with class "breadcrumb"

2. `/li[@class="active"]`
Find an `<li>` element with class "active" (the current page)

3. `/preceding-sibling::li[1]`
`preceding-sibling::` - Look at previous siblings (elements at the same level)

`li[1]` - Get the immediately preceding `<li>` sibling (the one right before the active one)

4. `/a/text()`
Get the text content of the `<a>` tag inside that `<li>`

`li[@class="active"]` = Finds the active li (position 4)

`preceding-sibling::li[1]` = Gets immediate previous sibling = Poetry (position 3)

`preceding-sibling::li[2]` = Gets second previous sibling = Books (position 2)

`preceding-sibling::li[3]` = Gets third previous sibling = Home (position 1)

1. Basic Selectors
`#` - ID Selector
Term: Hash/Octothorpe

Meaning: Selects element by ID

Example: #product_description → `<div id="product_description">`

`.` - Class Selector
Term: Dot/Period

Meaning: Selects elements by class

Example: .instock.availability → `<p class="instock availability">`

` ` (space) - Descendant Selector
Term: Descendant combinator

Meaning: Selects nested elements

Example: ul.breadcrumb li → `<ul><li>...</li></ul>`

2. Combinator Symbols
`>` - Child Selector
Term: Child combinator

Meaning: Direct children only (not grandchildren)

Example: ul > li → `<ul><li>Direct child</li></ul>`

`+` - Adjacent Sibling
Term: Next-sibling combinator

Meaning: Immediately following sibling

Example: th + td → `<th></th><td>Adjacent</td>`

`~` - General Sibling
Term: Subsequent-sibling combinator

Meaning: All following siblings

Example: h2 ~ p → `<h2></h2><p>Sibling 1</p><p>Sibling 2</p>`

3. Attribute Selectors
`[attr]` - Attribute Presence
Example: [href] → `<a href="...">`

`[attr=value]` - Exact Match
Example: [class="active"] → <li class="active">

`[attr^=value]` - Starts With
Example: [href^="https"] → Links starting with "https"

`[attr$=value]` - Ends With
Example: [src$=".jpg"] → Images ending with ".jpg"

`[attr*=value]` - Contains
Example: [class*="stock"] → Classes containing "stock"

4. Pseudo-classes (state/position)
`:contains(text)`
Scrapy-specific: Elements containing text

Example: th:contains("UPC") → <th>UPC</th>

`:nth-child(n)`
Term: Positional pseudo-class

Example: li:nth-child(3) → Third <li> element

`:first-child`, `:last-child`
Example: tr:first-child → First row in table

`:not(selector)`
Term: Negation pseudo-class

Example: div:not(.hidden) → Divs without class "hidden"

5. Pseudo-elements (content extraction)
`::text`
Scrapy-specific: Extract text content

Example: a::text → Text inside `<a>` tags

`::attr(attribute)`
Scrapy-specific: Extract attribute value

Example: a::attr(href) → URL from href attribute

6. Special Characters
`*` - Universal Selector
Meaning: Selects all elements

Example: div * → All elements inside div

`|` - Namespace separator
Rarely used: For XML namespaces

Key Terms:
Selector: Pattern to match elements

Combinator: Symbol defining relationship between selectors

Pseudo-class: Selects elements based on state/position

Pseudo-element: Accesses parts of elements/content

Specificity: How specific a selector is (IDs > classes > elements)

These symbols and terms form the language of CSS selection that makes web scraping precise and efficient!
