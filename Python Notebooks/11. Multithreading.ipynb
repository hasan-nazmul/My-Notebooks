```python
from threading import *

def display():
    for i in range(65, 91):
        print(chr(i))

t = Thread(target=display, name='Alphabets')

t.start()

for i in range(65, 91):
    print(i)

t.join()
```

```python
from threading import *

class Alphabets(Thread):
    def run(self):
        for i in range(65, 91):
            print(chr(i))
            sleep(0.1)

t = Alphabets()

t.start()

for i in range(65, 91):
    print(i)

t.join()
```

Make a class inherited from Thread and override the run method to display alphabets. And also make the task function named `run`.

```python
from threading import *
from time import *

class MyData():
    def __init__(self):
        self.data = 1
        self.flag = False
        self.lock = Lock()

    def puts(self, data):
        while self.flag:
            pass
        self.lock.acquire()
        self.data = data
        self.flag = True
        print('Producer:', data)
        self.lock.release()

    def gets(self):
        while not self.flag:
            pass
        self.lock.acquire()
        x = self.data
        self.flag = False
        self.lock.release()
        return x

def producer(data):
    a = 1
    b = 1
    n = 1
    while True:
        data.puts(n)
        a, b = b, n
        n = a+b
        sleep(1)

def consumer(data):
    while True:
        print('Consumer:', data.gets())
        sleep(1)

md = MyData()

t1 = Thread(target=lambda: producer(md))
t2 = Thread(target=lambda: consumer(md))

t1.start()
t2.start()

t1.join()
t2.join()
```

```python
from threading import *
from time import *

class MyData():
    def __init__(self):
        self.data = 1
        self.cv = Condition()

    def puts(self, data):
        self.cv.acquire()
        self.cv.wait(timeout=0)
        self.data = data
        print('Producer:', data)
        self.cv.notify()
        self.cv.release()

    def gets(self):
        self.cv.acquire()
        self.cv.wait(timeout=0)
        x = self.data
        self.cv.notify()
        self.cv.release()
        return x

def producer(data):
    a = 1
    b = 1
    n = 1
    while True:
        data.puts(n)
        a, b = b, n
        n = a+b
        sleep(1)

def consumer(data):
    while True:
        print('Consumer:', data.gets())
        sleep(1)

md = MyData()

t1 = Thread(target=lambda: producer(md))
t2 = Thread(target=lambda: consumer(md))

t1.start()
t2.start()

t1.join()
t2.join()
```

Both code works same, but here producer writes and notifies to the consumer, waits until cosuming is done. And then, consumer consumes and notifies the producer about ending consumption.

`timedelta` is basically duration.

```python
from datetime import *

year = 2021

count = 0

for month in range(1, 13):
    dt = date(year, month, 1)

    if dt.weekday() == 0:
        print(month)
        count += 1

print('Months that start from Mondays in the year of', year, ":", count)
```

```python
import datetime as dt

week_days = ['monday', 'tuesday', 'wednesday', 'thursday', 'friday', 'saturday', 'sunday']

def prev_date(day):
    today = dt.date.today()
    t_dw = today.weekday()
    dw = week_days.index(day.lower())

    diff = t_dw - dw

    if diff < 0:
        new_dt = today - dt.timedelta(7 + diff)

    else:
        new_dt = today - dt.timedelta(diff)

    return new_dt.strftime('%Y-%m-%d')

print(prev_date('saturday'))
```

```python
import datetime as dt

def all_sundays(year):
    dw = dt.date(year, 1, 1).weekday()
    first_sunday = dt.datetime(year, 1, 1) - dt.timedelta(dw-6)

    while True:
        print(first_sunday)
        first_sunday += dt.timedelta(7)
        if first_sunday.year > year:
            break

all_sundays(2022)
```
