```python
class Rectangle:
    count = 0
    def __init__(self, length=1, breadth=1):
        if not (isinstance(length, (int, float)) and isinstance(breadth, (int, float))):
            raise TypeError('Dimensions must be of numeric types.')
        if length < 0 or breadth < 0:
            raise ValueError('Dimensions has to be non-negative values!')
        self.length = length
        self.breadth = breadth
        Rectangle.count += 1

    def __str__(self):
        return f'Length: {self.length}\nBreadth: {self.breadth}'

    def area(self):
        return self.length * self.breadth

    def perimeter(self):
        return 2 * (self.length+self.breadth)

    def get_length_n_breadth(self):
        return (id(self.length), id(self.breadth))

    @classmethod
    def get_count(cls):
        return id(cls.count)

r = Rectangle(10, 25)
r1 = Rectangle(35, 40)
r2 = Rectangle(50, 10)
print(Rectangle.get_count(), r.get_count(), r1.get_count(), r2.get_count())
print(r.get_length_n_breadth())
print(r1.get_length_n_breadth())
print(r2.get_length_n_breadth())
```

But `length`, `breadth` or the methods: `area(self)`, `perimeter(self)` are instance variables and instance methods. Means each object will have its own `length`, `breadth` and `area(self)`, `perimeter(self)`. Thatswhy its possible to write `Rectangle.get_count()` not `Rectangle.area()` or `Rectangle.perimeter()`. `@classmethod` decorator needs to be mentioned before every classmethods each. In `get_count(cls)`, `cls` can have a class as a parameter.

```python
class MinimumBalanceError(Exception):
    pass

class Account:
    acc_number = 1001

    def __init__(self, name='John Doe', balance=5000):
        self.name = name
        self.acc = Account.acc_number

        try:
            self.balance = balance
        except Exception as e:
            print(e)
        else:
            print('Account has been created successfully!')

        Account.acc_number += 1

    @property
    def balance(self):
        return self._balance

    @balance.setter
    def balance(self, balance):
        if not isinstance(balance, (int, float)):
            raise ValueError('Balance has to be of numeric value!')
        if balance < 5000:
            raise MinimumBalanceError('Balance cannot be lesser than 5000!')
        self._balance = balance

    def deposit(self, amount):
        try:
            self.balance += amount
        except Exception as e:
            print(e)
            print(f'Deposit was unsuccessful!')
        else:
            print(f'Deposit of ${amount} to {self.name} was successful!')

    def withdraw(self, amount):
        try:
            self.balance -= amount
        except Exception as e:
            print(e)
            print(f'Withdrawal of {amount} was unsuccessful!')
        else:
            print(f'Withdrawal of ${amount} from {self.name} was successful!')

    def __str__(self):
        return f'Account No: {self.acc}, Name: {self.name}, Balance: ${self.balance}'

a = Account('Nazmul', 12000)
a.deposit(7000)
print(a)
```

```python
import math
from abc import ABC, abstractmethod

class TriangleFormationError(Exception):
    pass

class Polygon(ABC):
    def __init__(self, no_of_sides, *sides):
        self.no_of_sides = no_of_sides
        self.sides = sides
    @abstractmethod
    def Area(self):
        pass

class Triangle(Polygon):
    def __init__(self, *sides):
        super().__init__(3, *sides)
    def Area(self):
        s = sum(self.sides) / 2
        area = math.sqrt(s * math.prod([(s-x) for x in self.sides]))
        return area
    @property
    def sides(self):
        return self._sides
    @sides.setter
    def sides(self, sides):
        if len(sides) != 3:
            raise TriangleFormationError('Triangle must have exactly 3 sides!')
        s = sum(sides)
        for x in sides:
            if not isinstance(x, (int, float)) or x <= 0 or s-2*x < 0:
                raise TriangleFormationError('Triangle Forming is not possible with these sides!')
        self._sides = sides

try:
    t = Triangle(5, 4, 10)
    print(t.sides)
except Exception as e:
    print(e)

try:
    t = Triangle(3, 5, 4)
    print(t.sides)
    print(t.Area())
except Exception as e:
    print(e)
```

```python
import math

class Rational:

    def __init__(self, p, q):

        self.p = p // math.gcd(p, q)
        self.q = q // math.gcd(p, q)

    def __add__(self, other):
        p = self.p * other.q + other.p * self.q
        q = self.q * other.q
        return Rational(p, q)

    def __sub__(self, other):
        p = self.p * other.q - other.p * self.q
        q = self.q * other.q
        return Rational(p, q)

    def __str__(self):
        return f'{self.p}/{self.q}'

r1 = Rational(10, 15)
r2 = Rational(9, 3)

print(r1, '+', r2, '=', r1+r2)
print(r1, '-', r2, '=', r1-r2)

```

```python
print("\n=== Common OOP Pitfalls ===\n")

# ❌ PITFALL 1: Mutable class attributes
print("❌ Pitfall 1: Shared Mutable Class Attributes")

class BadGame:
    players = []  # Shared across all instances!
    
    def __init__(self, name):
        self.name = name
        self.players.append(name)

game1 = BadGame("Alice")
game2 = BadGame("Bob")

print(f"  Game1 players: {game1.players}")
print(f"  Game2 players: {game2.players}")
print("  ✗ Both games share same player list!")

# ✓ SOLUTION
print("\n✓ Solution: Use instance attributes")

class GoodGame:
    def __init__(self, name):
        self.name = name
        self.players = []  # Instance-specific
    
    def add_player(self, player):
        self.players.append(player)

game1 = GoodGame("Game1")
game2 = GoodGame("Game2")

game1.add_player("Alice")
game2.add_player("Bob")

print(f"  Game1 players: {game1.players}")
print(f"  Game2 players: {game2.players}")
print("  ✓ Each game has its own players")

# ❌ PITFALL 2: Broken inheritance
print("\n\n❌ Pitfall 2: Inappropriate Inheritance")

class Rectangle:
    def __init__(self, width, height):
        self.width = width
        self.height = height

class Square(Rectangle):  # Is Square really a Rectangle?
    def __init__(self, side):
        super().__init__(side, side)
    
    def set_width(self, width):
        self.width = width
        self.height = width  # Violates Rectangle contract

print("  Square inherits from Rectangle")
print("  ✗ But Square::set_width breaks Rectangle behavior!")

# ✓ SOLUTION
print("\n✓ Solution: Composition over Inheritance")

class Shape:
    pass

class RectangleShape(Shape):
    def __init__(self, width, height):
        self.width = width
        self.height = height

class SquareShape(Shape):
    def __init__(self, side):
        self.side = side

print("  Use separate classes or shared base")
print("  ✓ No broken contracts!")
```

```python
print("\n=== Lazy Loading Optimization ===\n")

class ExpensiveResource:
    def __init__(self):
        print("  Expensive resource loaded (slow!)")
        self.data = "large_dataset_" * 1000

class User:
    def __init__(self, user_id):
        self.user_id = user_id
        self._profile = None  # Lazy load
        self._resource = None
    
    @property
    def profile(self):
        if self._profile is None:
            print(f"  Loading profile for user {self.user_id}")
            self._profile = {"name": "Alice", "email": "alice@example.com"}
        return self._profile
    
    @property
    def resource(self):
        if self._resource is None:
            self._resource = ExpensiveResource()
        return self._resource

print("Lazy Loading Pattern:")
user = User(123)
print("  User created (no loading yet)")

print("\nAccessing profile:")
profile = user.profile
print(f"  Profile: {profile}")

print("\nAccessing resource:")
resource = user.resource
print(f"  Resource loaded")
```

```python
print("\n=== Observer Pattern ===\n")

class Subject:
    """Observable object"""
    def __init__(self):
        self._observers = []
        self._state = None
    
    def attach(self, observer):
        if observer not in self._observers:
            self._observers.append(observer)
    
    def detach(self, observer):
        self._observers.remove(observer)
    
    def notify(self):
        for observer in self._observers:
            observer.update(self._state)
    
    @property
    def state(self):
        return self._state
    
    @state.setter
    def state(self, value):
        self._state = value
        self.notify()

class Observer(ABC):
    @abstractmethod
    def update(self, state):
        pass

class UserInterface(Observer):
    def update(self, state):
        print(f"  [UI] Updated: {state}")

class Logger(Observer):
    def update(self, state):
        print(f"  [LOG] State changed to: {state}")

# Usage
subject = Subject()
ui = UserInterface()
logger = Logger()

subject.attach(ui)
subject.attach(logger)

print("Observer Pattern - State Updates")
subject.state = "User logged in"
subject.state = "Processing payment"
```

```python
print("\n=== Factory Pattern ===\n")

from abc import ABC, abstractmethod

# Abstract base
class Payment(ABC):
    @abstractmethod
    def process(self, amount):
        pass

# Concrete implementations
class CreditCardPayment(Payment):
    def process(self, amount):
        return f"Processing ${amount} via Credit Card"

class PayPalPayment(Payment):
    def process(self, amount):
        return f"Processing ${amount} via PayPal"

class CryptoPayment(Payment):
    def process(self, amount):
        return f"Processing ${amount} via Crypto"

# Factory
class PaymentFactory:
    @staticmethod
    def create_payment(payment_type):
        payments = {
            'card': CreditCardPayment(),
            'paypal': PayPalPayment(),
            'crypto': CryptoPayment(),
        }
        return payments.get(payment_type)

# Usage
print("Factory Pattern - Payment Processing")
for method in ['card', 'paypal', 'crypto']:
    payment = PaymentFactory.create_payment(method)
    print(f"  {payment.process(99.99)}")
```

```python
print("=== Singleton Pattern ===\n")

# Singleton using decorator
def singleton(cls):
    instances = {}
    def get_instance(*args, **kwargs):
        if cls not in instances:
            instances[cls] = cls(*args, **kwargs)
        return instances[cls]
    return get_instance

@singleton
class DatabaseConnection:
    """Only one instance of database connection"""
    def __init__(self):
        self.connection_id = id(self)
        print(f"  Creating DatabaseConnection (id: {self.connection_id})")
    
    def query(self, sql):
        return f"Executing: {sql}"

# Test singleton
db1 = DatabaseConnection()
db2 = DatabaseConnection()
print(f"\ndb1 is db2: {db1 is db2}  (Same instance: ✓)")

# Metaclass approach
class SingletonMeta(type):
    _instances = {}
    
    def __call__(cls, *args, **kwargs):
        if cls not in cls._instances:
            cls._instances[cls] = super().__call__(*args, **kwargs)
        return cls._instances[cls]

class Logger(metaclass=SingletonMeta):
    def __init__(self):
        self.logs = []
        print("  Logger initialized")
    
    def log(self, message):
        self.logs.append(message)
        print(f"  [LOG] {message}")

logger1 = Logger()
logger2 = Logger()
print(f"logger1 is logger2: {logger1 is logger2}  (Same instance: ✓)")
```
