> Database Connectivity

```python
import sqlite3

conn = sqlite3.connect('shop.db')

curr = conn.cursor()

def write(table_name, data):
    for dt in data:
        s = f'insert into {table_name} values('
        for d in dt:
            if not isinstance(d, str):
                s += f'{d},'
            else:
                s += f'\"{d}\",'
        s = s[:-1]
        s += ')'
        curr.execute(s)
```

All queries will be executed by the cursor, cursor works through the connection.

```python
customer_data = [
    (101, "Anita", "Delhi"),
    (102, "Raj", "Hyperabad"),
    (103, "Mahwid", "Kolkata"),
    (107, "Al", "Delhi"),
    (109, "Sharma", "Chennai")
]

product_data = [
    (10, "Toothpaste", 20),
    (11, "Toothbrush", 10),
    (12, "Lotion", 30),
    (13, "Shampoo", 25),
    (14, "Scap", 10)
]

order_data = [
    (10005, 101, 10, 1),
    (10005, 101, 11, 2),
    (12389, 109, 12, 1),
    (12389, 109, 14, 1),
    (12389, 109, 11, 2),
    (63017, 103, 13, 4),
    (63017, 103, 10, 1),
    (63017, 103, 11, 4),
    (71222, 101, 12, 2),
    (71222, 101, 13, 1),
    (96351, 102, 14, 1)
]
```

Creating Tables

```python
curr.execute('create table if not exists Customer(id int primary key, name text, city text)')
curr.execute('create table if not exists Product(id int primary key, name text, price int)')
curr.execute('create table if not exists Orders(id int, customer_id int, product_id int, quantity int, foreign key(customer_id) references Customer(id), foreign key(product_id) references Product(id))')
```

```python
write('Customer', customer_data)
write('Product', product_data)
write('Orders', order_data)
```

Have to commit the changes.

```python
conn.commit()
```

```python
rows = curr.execute('select O.id,C.name,P.name,P.price,O.quantity from Orders O,Customer C,Product P where O.customer_id=C.id and O.product_id=P.id').fetchall()

for row in rows:
    print(row)
```

```python
rows = curr.execute('select name,sum(quantity) from Orders,Product where Orders.product_id=Product.id group by Product.name').fetchall()

for row in rows:
    print(row)
```

```python
curr.close()
conn.close()
```

```python
import sqlite3
import datetime

# Connect to the database with type detection enabled
conn = sqlite3.connect(':memory:', detect_types=sqlite3.PARSE_DECLTYPES | sqlite3.PARSE_COLNAMES)
cursor = conn.cursor()

# Create a table with a DATETIME column
cursor.execute('''
    CREATE TABLE events (
        id INTEGER PRIMARY KEY,
        event_time DATETIME
    )
''')

# Insert a Python datetime object
now = datetime.datetime.now()
cursor.execute("INSERT INTO events (event_time) VALUES (?)", (now,))
conn.commit()

# Retrieve the datetime object
cursor.execute("SELECT event_time FROM events WHERE id = 1")
retrieved_time = cursor.fetchone()[0]

print(f"Original datetime: {now}")
print(f"Retrieved datetime: {retrieved_time}")
print(f"Type of retrieved datetime: {type(retrieved_time)}")

conn.close()
```
