> csv module

```python
import csv
```

```python
f = open('names.csv', 'r')  # Open a file for reading
csv_reader = csv.reader(f)  # Create a CSV reader object
```

```python
for row in csv_reader:
    print(row)  # Print each row in the CSV file
```

```python
f.close()  # Close the file after reading
```

```python
f = open('names.csv', 'r')  # Reopen the file for reading
```

```python
rdr = csv.DictReader(f)  # Create a CSV DictReader object
```

```python
import pprint

data = {}

id = 1

for row in rdr:
    data[id] = row  # Store each row in a dictionary with 'Name' as the key
    id += 1

pprint.pprint(data)  # Pretty print the dictionary
```

```python
covid_data = [
    ('Country', 'Doses', 'People', 'Percentage'),
    ('India', '38Cr', '84.1Cr', 61),
    ('China', '38Cr', '12Cr', 88.1),
    ('United States', '56.8Cr', '21.9Cr', 66.4),
    ('Brazil', '42.4Cr', '16.2Cr', 76.4),
    ('Indonesia', '39Cr', '16.2Cr', 59.3)
]

f = open('covid_data.csv', 'w', newline='')  # Open a file for writing

csv_writer = csv.writer(f)  # Create a CSV writer object

csv_writer.writerows(covid_data)  # Write the rows to the CSV file

f.close()  # Close the file after writing
```

```python
import csv

covid = [
    {'Country': 'India', 'Doses': '186Cr', 'People': '84.1Cr', 'Percentage': 61},
    {'Country': 'China', 'Doses': '330Cr', 'People': '124Cr', 'Percentage': 88.1},
    {'Country': 'United States', 'Doses': '56.8Cr', 'People': '21.9Cr', 'Percentage': 66.4},
    {'Country': 'Brazil', 'Doses': '42.4Cr', 'People': '16.2Cr', 'Percentage': 76.4},
    {'Country': 'Indonesia', 'Doses': '39Cr', 'People': '16.2Cr', 'Percentage': 59.3}
]

f = open('covid_data_dict.csv', 'w', newline='')  # Open a file for writing
csv_writer = csv.DictWriter(f, fieldnames=covid[0].keys())  # Create a CSV DictWriter object
csv_writer.writeheader()  # Write the header row
csv_writer.writerows(covid)  # Write the rows to the CSV file   
f.close()  # Close the file after writing
```
